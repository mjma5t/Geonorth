```python
import zipfile

with zipfile.ZipFile(f"era5_land_{vari}_2005-15.nc", "r") as z:
    print(z.namelist())  # see what's inside
    z.extractall("era5_west_2")  # extract to a folder
```

```python
import xarray as xr
era5 = xr.open_dataset("era5_west_2/data_0.nc")

era5
```

```python

sites = {
    "TrailValleyCreek":   {"lat": 68.7462,    "lon": -133.5017 },
    "HavikpakCreek":      {"lat": 68.3203,    "lon": -133.5188},
    "SmithCreek":         {"lat": 63.1534,    "lon": -123.2522},
    "BakerCreek":         {"lat": 62.715894,  "lon": -114.574885},
}

era5_sites = {}
for site_name, coords in sites.items():
    era5_sites[site_name] = era5[f"{abbr}"].sel(
        latitude=coords["lat"],
        longitude=coords["lon"],
        method="nearest"
    )

# Verify nearest grid cell coordinates for each site
for site_name, da in era5_sites.items():
    print(f"{site_name}: latitude={float(da.latitude):.4f}, longitude={float(da.longitude):.4f}")



```

```python

# Load and combine the two ERA5 files
files = ["era5_west/data_0.nc", "era5_west_2/data_0.nc"]
era5 = xr.open_mfdataset(files, combine="by_coords")

# Extract Smith Creek
era5_sc = era5[f"{abbr}"].sel(
     latitude=63.1534,
     longitude=-123.2522,
     method="nearest"
)

# Extract TVC
era5_tvc = era5[f"{abbr}"].sel(
     latitude=68.7462,
     longitude=-133.5017,
     method="nearest"
)

# Extract HC
era5_hc = era5[f"{abbr}"].sel(
     latitude=68.3203,
     longitude=-133.5188,
     method="nearest"
)

# Extract Baker Creek
era5_bc = era5[f"{abbr}"].sel(
     latitude=62.715894,
     longitude=-114.574885,
     method="nearest"
)

```
