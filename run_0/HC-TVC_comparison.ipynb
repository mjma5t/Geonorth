```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import pandas as pd
import os
```

```python
df1 = xr.open_dataset("HC_1995-2014.nc", engine="netcdf4")
df2 = xr.open_dataset("TVC_1995-2014.nc", engine="netcdf4")
```

```python
# Snow Depth

var_HC = df1["SNOW_DEPTH"]
var_HC
var_TVC = df2["SNOW_DEPTH"]
var_TVC

var_HC.plot()
var_TVC.plot()
```

```python
# Snow Depth

var_HC = df1["H2OSOI"]
var_HC_1 = var_HC.isel(levsoi=1)

var_TVC = df2["H2OSOI"]
var_TVC_1 = var_TVC.isel(levsoi=1)

var_HC_1.plot()
var_TVC_1.plot()
```
